## 3. Install Android Command Line Tools

```python
import os

# 1. Rudisha working directory kwenye root (/content) ili kuondoa getcwd errors
os.chdir('/content')

# 2. Wezesha Environment Variables za Android SDK
os.environ['ANDROID_HOME'] = '/content/android-sdk'
os.environ['ANDROID_SDK_ROOT'] = '/content/android-sdk'
os.environ['PATH'] = os.environ['PATH'] + ':/content/android-sdk/cmdline-tools/latest/bin:/content/android-sdk/platform-tools:/content/android-sdk/build-tools/34.0.0'

# 3. Pakua na ku-extract Android Command Line Tools
!mkdir -p /content/android-sdk/cmdline-tools
!wget -q https://dl.google.com/android/repository/commandlinetools-linux-11076708_latest.zip -O cmdtools.zip
!unzip -q -o cmdtools.zip -d android-sdk/cmdline-tools
!rm -rf /content/android-sdk/cmdline-tools/latest
!mv /content/android-sdk/cmdline-tools/cmdline-tools /content/android-sdk/cmdline-tools/latest

# 4. Kubali Licenses na weka Android Packages
!yes | sdkmanager --licenses > /dev/null 2>&1 || true
!sdkmanager "platform-tools" "platforms;android-34" "build-tools;34.0.0" "platforms;android-29" > /dev/null 2>&1

print('Android SDK ready')
```

## 4. Upload KaloriAI-Final.zip

Upload faili **KaloriAI-Final.zip** hapa (au weka link ya download).

```python
from google.colab import files
import zipfile, os, shutil

print('Upload KaloriAI-APK-FIX.zip ...')
uploaded = files.upload()

zip_name = list(uploaded.keys())[0]
print('Got:', zip_name)

# Hakikisha faili ipo
if not os.path.exists(zip_name):
    zip_name = '/content/' + zip_name

print('Extracting', zip_name)

if os.path.exists('/content/kalorai-full'):
    shutil.rmtree('/content/kalorai-full')

with zipfile.ZipFile(zip_name, 'r') as z:
    z.extractall('/content')

project = '/content/kalorai-full'
if not os.path.exists(project):
    for root, dirs, files_ in os.walk('/content'):
        if 'package.json' in files_ and 'public' in dirs:
            project = root
            break

print('Project path:', project)
!ls -la {project}
```

## 5. Install npm packages + Capacitor Android

```python
import os
os.chdir(project)
print('Working in:', os.getcwd())

!npm install --legacy-peer-deps 2>&1 | tail -20

# Add Android platform if not exists
!npx cap add android 2>&1 | tail -15 || true
!npx cap sync android 2>&1 | tail -20

print('Capacitor Android ready')
!ls android/app 2>/dev/null || ls -la
```

## 6. Fix Gradle + Build APK

```python
import os

# Set variable ya project path (iliyo sahihi baada ya ku-extract zip)
project = '/content/KaloriAI-Android-User'

# Lazimisha mfumo kutumia Java 17 badala ya Java 21
os.environ['JAVA_HOME'] = '/usr/lib/jvm/java-17-openjdk-amd64'
os.environ['PATH'] = '/usr/lib/jvm/java-17-openjdk-amd64/bin:' + os.environ['PATH']

os.chdir(project)

# Hakikisha local.properties ina njia sahihi ya Android SDK
local_props = os.path.join(project, 'android', 'local.properties')
os.makedirs(os.path.dirname(local_props), exist_ok=True)
with open(local_props, 'w') as f:
    f.write('sdk.dir=/content/android-sdk\n')

# Build debug APK ukitumia Java 17
os.chdir(os.path.join(project, 'android'))
print('Building APK... this can take 5-10 minutes')
!chmod +x gradlew 2>/dev/null || true
!JAVA_HOME=/usr/lib/jvm/java-17-openjdk-amd64 ./gradlew assembleDebug --no-daemon

print('\n--- Looking for APK ---')
!find . -name "*.apk" 2>/dev/null
```

## 7. Download APK

```python
from google.colab import files
import glob, os

apks = glob.glob(os.path.join(project, 'android/**/*.apk'), recursive=True)
apks = [a for a in apks if 'app-debug' in a or a.endswith('.apk')]

if not apks:
    # search wider
    apks = glob.glob('/content/**/*.apk', recursive=True)

print('Found APKs:', apks)

if apks:
    apk_path = apks[0]
    print('Downloading:', apk_path)
    files.download(apk_path)
    print('DONE! Install on Pixel 6a (enable Install unknown apps)')
else:
    print('APK not found. Check build logs above for errors.')
    print('Common fix: re-run cell 6, or use PWA instead (Add to Home Screen).')
```
